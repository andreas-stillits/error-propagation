```python
from my_package import propagate, Constant, Empirical, Model, Normal
import numpy as np
```

```python
rng = np.random.default_rng(seed=42)
rad_per_deg = np.pi / 180.0

theta = Normal(35*rad_per_deg, 1*rad_per_deg)
v = Normal(10, 0.1)
g = Constant(9.82)

def get_range(theta: float, v: float, g: float) -> float:
    return v**2 * np.sin(2*theta) / g 

results = propagate(rng, get_range, inputs=[theta, v, g], n_samples=10_000)
print(f"The estimated mean range is {np.mean(results):.2f} m with a standard deviation of {np.std(results):.2f} m.")
```
